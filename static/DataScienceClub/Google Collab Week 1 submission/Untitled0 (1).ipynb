```python
import seaborn as sns
import matplotlib.pyplot as plt
data = sns.load_dataset("tips")
data.head()
data.tail()

```

```python
sns.barplot(x = "sex", y = "tip", data = data)
plt.title("i am good", fontsize = 20)
plt.ylabel("i am good", fontsize = 20)
plt.xlabel("i am good", fontsize = 20)
```

```python
sns.lineplot(x = "sex", y = "tip", data = data)
plt.title("i am good", fontsize = 20)
plt.xlabel("i am good", fontsize = 20)
plt.ylabel("i am good", fontsize = 20)
```

```python
sns.scatterplot(x = "tip", y = "total_bill", data = data)
plt.xlabel("i am good", fontsize = 20)
plt.ylabel("i am good", fontsize = 20)
plt.title("i am good", fontsize = 20)
```

```python
sns.kdeplot(x="tip", data = data)
plt.title("i am good", fontsize = 20)
plt.xlabel("i am good", fontsize = 20)
```

```python
sns.histplot(x="tip", data = data)
plt.title("i am good")
plt.ylabel("i am good", fontsize = 20)
plt.xlabel("i am good", fontsize = 20)
```

```python
matrix = data[["tip","size","total_bill"]].corr()
sns.heatmap(matrix, annot = True)
plt.title("i am good", fontsize = 20)
```
