```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some sample data
x_data = np.random.rand(100)
y_data = 2 * x_data + 1 + np.random.randn(100) * 0.2

# Create a scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(x_data, y_data, color='green', label='Random Data')
plt.title('Custom Scatter Plot')
plt.xlabel('Feature X')
plt.ylabel('Feature Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

# Reshape x_data for sklearn (needs to be 2D array)
X = x_data.reshape(-1, 1)

# Create a linear regression model
regr = LinearRegression()

# Train the model using the training sets
regr.fit(X, y_data)

# Make predictions using the regression model
y_pred = regr.predict(X)

# Create a scatter plot with the regression line
plt.figure(figsize=(8, 6))
plt.scatter(x_data, y_data, color='green', label='Random Data')
plt.plot(x_data, y_pred, color='red', linewidth=2, label='Regression Line')
plt.title('Custom Scatter Plot with Regression Line')
plt.xlabel('Feature X')
plt.ylabel('Feature Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_data, y_pred)
print(f"Mean Squared Error: {mse}")
```

```python
print(f"Intercept: {regr.intercept_}")
print(f"Slope: {regr.coef_[0]}")
print(f"R2 Score: {r2_score(y_data, y_pred)}")
```
