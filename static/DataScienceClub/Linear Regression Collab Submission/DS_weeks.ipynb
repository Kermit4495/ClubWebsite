```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset("tips")
```

```python
tips.head()
```

```python
tips.tail()
```

```python
sns.lineplot(x="total_bill",y="tip",data=tips)
plt.title("Line Graph",fontsize=32)
```

```python
sns.barplot(x="sex",y="tip",data=tips)
plt.title("Bar Graph",fontsize=32)
plt.xlabel("sex",fontsize=15)
```

```python
corr=tips[["total_bill","tip","size"]].corr()
sns.heatmap(corr,annot=True)
plt.title("Heatmap",fontsize=32)
```

```python
sns.scatterplot(x="total_bill",y="tip",data=tips)
plt.title("Scatterplot",fontsize=32)
plt.ylabel("tips",fontsize=15)
plt.xlabel("total_bill",fontsize=15)
```

```python
sns.kdeplot(x="tip",data=tips)
plt.title("Density Plot",fontsize=32)
plt.xlabel("tips",fontsize=15)
plt.ylabel("density",fontsize=15)
```

```python
sns.histplot(x="tip",data=tips)
plt.title("Histogram",fontsize=32)
```
